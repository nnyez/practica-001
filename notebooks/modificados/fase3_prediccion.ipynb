## **1. Objetivos y alcance**

- Función que recibe `product_id` y una fecha y devuelve la predicción de demanda.
- Función que recibe un lote de productos para la misma fecha y devuelve la lista de predicciones.
- Usar el modelo LSTM entrenado en la fase anterior (`modelos/fase2/lstm_final.keras`).
- Código simple y estructurado, en el paquete `src/fase3/`.

**Flujo:**

```
producto_id + fecha
        │
        ▼
últimos 7 días de historial  →  ventana (7, 15)
        │
        ▼
escalar (0-1)  →  LSTM  →  desescalar  →  unidades reales
```


## **2. Setup del entorno**

````python
import os
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '3'  # silencia logs de TensorFlow

import sys
from pathlib import Path

BASE = Path('..').resolve()
if str(BASE) not in sys.path:
    sys.path.insert(0, str(BASE))

import pandas as pd

from src.fase3 import inicializar, predecir_batch, predecir_venta
from src.fase3.config import RUTA_DATOS
from src.fase3.entrada import construir_ventana, ventana_disponible

print('Entorno listo.')
````

## **3. Carga del modelo y el escalador**

- El LSTM se entrenó con las 15 variables escaladas entre 0 y 1.
- El escalador se carga desde `modelos/fase2/escalador.pkl` (guardado en el entrenamiento), no se reconstruye.
- La carga es automática la primera vez que se predice; aquí se hace explícita para dejarlo claro.

````python
inicializar()
print('Modelo y escalador cargados.')
print('Ventana esperada:', 7, 'días')
````

## **4. Construcción de la ventana de entrada**

Para predecir una fecha se toman los 7 días anteriores de ese producto. 

Cada fila contiene: `cantidad_vendida`, `es_fin_de_semana`, `es_feriado`, `dia_sin`, `dia_cos` y el one-hot del producto (15 columnas en total).

````python
df = pd.read_csv(RUTA_DATOS, encoding='utf-8-sig', parse_dates=['fecha_venta'])

producto, fecha = 'P0001', '2022-03-02'
print('¿Ventana disponible?', ventana_disponible(df, producto, fecha))

X = construir_ventana(df, producto, fecha)
print('Forma:', X.shape)
print('Primera fila:', X[0])
print('Última fila :', X[-1])
````

## **5. Predicción individual: `predecir_venta`**

La función predecir_venta recibe el identificador de un producto y una fecha, utiliza los últimos 7 días de historial y genera una predicción de la demanda.

 Finalmente, devuelve el resultado en unidades reales de venta.

````python
for producto in ['P0001', 'P0008']:
    print(f'{producto} 2022-03-02: {predecir_venta(producto, "2022-03-02"):.2f} unidades')
````

## **6. Predicción en lote: `predecir_batch`**

La función predecir_batch permite predecir las ventas de varios productos para una misma fecha. 

Si ocurre un error con algún producto, se registra sin interrumpir las predicciones de los demás.


````python
for r in predecir_batch(['P0001', 'P0002', 'P0003', 'P0008', 'P0010'], '2022-03-02'):
    print(r)
````

## **7. Comparación con valor real**

Se prueba la función predecir_venta con el producto P0001 y la fecha 2022-02-28. 

La predicción se compara con la venta real registrada para calcular la diferencia en unidades y el porcentaje de error, permitiendo evaluar qué tan cercana fue la estimación del modelo.

````python
fecha, producto = pd.Timestamp('2022-02-28'), 'P0001'

prediccion = predecir_venta(producto, fecha)
real = float(df.loc[(df['product_id'] == producto) & (df['fecha_venta'] == fecha), 'cantidad_vendida'].iloc[0])

print(f'{producto} {fecha.date()}')
print(f'  Predicción: {prediccion:.2f}')
print(f'  Real      : {real:.2f}')
print(f'  Error     : {abs(prediccion - real):.2f} unidades ({abs(prediccion - real) / real * 100:.1f}%)')
````

## **8. Casos de error**

- Producto inexistente.
- Fecha sin 7 días de historial previo.
- Lote mixto: los productos válidos se predicen y los inválidos se reportan.

````python
for producto, fecha in [('P9999', '2022-03-02'), ('P0001', '2022-01-05')]:
    try:
        predecir_venta(producto, fecha)
    except ValueError as exc:
        print(f'{producto} {fecha} -> {exc}')

print()
for r in predecir_batch(['P0001', 'P9999', 'P0002'], '2022-03-02'):
    print(r)
````

## **9. Conclusiones**

- Las funciones predecir_venta y predecir_batch permiten predecir las ventas de uno o varios productos.

- Se reutilizan el modelo LSTM y el escalador de la Fase 2, sin necesidad de volver a entrenarlos.

- La predicción por lote permite continuar con los demás productos si alguno presenta un error.

- Para obtener predicciones, se necesitan los datos reales de los 7 días anteriores de cada producto
